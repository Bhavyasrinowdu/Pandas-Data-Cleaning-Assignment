```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("data.csv")
print("\nShape:")
print(df.shape)
print("\nFirst 5 Rows:")
print(df.head())
print("\nDataset Information:")
df.info()
print("\nMissing Values Before Cleaning:")
print(df.isnull().sum())
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df["Calories"] = df["Calories"].fillna(df["Calories"].mean())
df = df.dropna(subset=["Date"])
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
filtered_data = df[df["Duration"] > 45]
print("\nFiltered Data:")
print(filtered_data)
df["Pulse_Difference"] = df["Maxpulse"] - df["Pulse"]
print("\nDataset After Creating New Column:")
print(df)
print("\nBasic Statistics:")
print(df.describe())
plt.figure(figsize=(8, 5))
plt.plot(df["Duration"])
plt.title("Exercise Duration")
plt.xlabel("Record")
plt.ylabel("Duration")
plt.show()
plt.figure(figsize=(8, 5))
sns.scatterplot(x="Duration", y="Calories", data=df)
plt.title("Duration vs Calories")
plt.xlabel("Duration")
plt.ylabel("Calories")
plt.show()
plt.figure(figsize=(8, 5))
sns.histplot(df["Pulse"], bins=10)
plt.title("Pulse Distribution")
plt.xlabel("Pulse")
plt.ylabel("Count")
plt.show()
df.to_csv("cleaned_data.csv", index=False)
print("\nAssignment completed successfully!")
print("Cleaned dataset saved as cleaned_data.csv")
```
